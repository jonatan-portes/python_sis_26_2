```python
with open("log.txt", "r") as log:
    conteudo = log.read()
    print(conteudo)
```

```python
with open("log.txt", "w") as log:
  log.write("""2026-10-14 18:42:00 - INFO - Nova informação adicionada""")
```

```python
with open("log.txt", "a+") as log:
  log.write ("""2026-10-14 08:30:15 - INFO - Sistema inicializado com sucesso
2026-10-14 08:31:02 - INFO - Usuario 'admin' realizou login
2026-10-14 08:35:22 - AVISO - Tentativa de acesso com senha incorreta: usuario 'bat'
2026-10-14 08:35:40 - SUCESSO - Usuario 'bat' realizou login
2026-10-14 08:40:11 - ERRO - Falha ao conectar ao banco de dados: timeout
2026-10-14 08:42:00 - INFO - Conexao reestabelecida
""")
  log.seek(0)
  conteudo = log.read()
  print(conteudo)
```

Métodos de Leitura


readline()

```python
with open("log.txt", "r", encoding="utf-8") as log:
    primeira_linha = log.readline()
    segunda_linha = log.readline()

print("1ª chamada:", primeira_linha.strip())
print("2ª chamada:", segunda_linha.strip())
```

file.readlines()

```python
with open("log.txt", "r", encoding="utf-8") as log:
    linhas = log.readlines()

print("Tipo:", type(linhas))
print("Quantidade de linhas:", len(linhas))
print("Última linha:", linhas[-1].strip())
print("Lista completa:", linhas)
```
