# Arquivos **CSV**

```python
#Exemplo de Escrita em CSV:
import csv
with open("pets.csv", "w", newline="") as file:
 writer = csv.writer(file)
 writer.writerow(["Nome", "Espécie", "Idade"])
 writer.writerow(["Bat", "Gato", "8"])
 writer.writerow(["Chai", "Cão", "10"])
```

```python
#Exemplo de Leitura em CSV:
import csv
with open("pets.csv", "r") as file:
 reader = csv.reader(file)
 for row in reader:
  print(row)
print("Leitura concluída!")
print(type(row))
print(type(reader))
```

```python
import csv

# 1. Escrevendo dados tabulares em CSV
servidores = [
    ["Host", "IP", "Status"],
    ["srv-app-01", "192.168.1.10", "Ativo"],
    ["srv-db-01", "192.168.1.20", "Manutencao"],
    ["srv-web-01", "192.168.1.30", "Ativo"]
]

with open("servidores.csv", "w", newline="", encoding="utf-8") as arq_csv:
    escritor = csv.writer(arq_csv, delimiter=";")
    escritor.writerows(servidores)

print("Arquivo servidores.csv criado com sucesso!")

# 2. Lendo o CSV linha a linha
with open("servidores.csv", "r", encoding="utf-8") as arq_csv:
    leitor = csv.reader(arq_csv, delimiter=";")
    for linha in leitor:
        print(linha)
```

# Arquivos **JSON**

```python
import json

config = {
    "ambiente": "Produção",
    "backup_ativo": True,
    "portas": [80, 443, 22]
}

# Salvando no arquivo
with open("config.json", "w", encoding="utf-8") as f:
    json.dump(config, f, indent=4)

# Lendo do arquivo
with open("config.json", "r", encoding="utf-8") as f:
    dados_lidos = json.load(f)
    print(dados_lidos["ambiente"])
```

```python
import json

pets = {
  "pets": [
    {"nome": "Bat", "especie": "gato", "comidas_favoritas": ["frango", "milho"]},
    {"nome": "Chai", "especie": "cachorro", "comidas_favoritas": ["carne", "biscoito"]}
  ]
}

# Salvando no arquivo
with open("pets.json", "w", encoding="utf-8") as f:
    json.dump(pets, f, indent=4)

# Lendo do arquivo
with open("pets.json", "r", encoding="utf-8") as f:
    dados_lidos = json.load(f)
    print(dados_lidos["pets"][0]["nome"])
```

# Planilhas

Exemplo básico

```python
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

wb = Workbook()
ws = wb.active
ws.title = "Servidores"

# Inserindo cabeçalho e dados
ws.append(["Hostname", "IP", "Uso de Disco (%)"])
ws.append(["srv-web-01", "192.168.1.10", 85])

# Destacando célula se uso for crítico
for linha in range(2, ws.max_row + 1):
    celula_disco = ws.cell(row=linha, column=3)
    if celula_disco.value >= 80:
        celula_disco.fill = PatternFill(
            start_color="FF9999",
            end_color="FF9999",
            fill_type="solid"
        )

wb.save("inventario.xlsx")
```

Escrita de dados

```python
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

# Criação da pasta de trabalho e seleção da aba ativa
wb = Workbook()
ws = wb.active
ws.title = "Relatorio_TI"

# Inserindo o cabeçalho
ws.append(["ID Chamado", "Sistema", "Prioridade", "Tempo Atendimento (h)"])

# Inserindo registros
chamados = [
    [101, "ERP", "Alta", 3.5],
    [102, "E-mail", "Baixa", 0.5],
    [103, "Banco de Dados", "Crítica", 5.0],
    [104, "Firewall", "Média", 2.0]
]

for linha in chamados:
    ws.append(linha)

# Estilizando o cabeçalho (Negrito e Fundo Cinza)
for celula in ws[1]:
    celula.font = Font(bold=True, color="FFFFFF")
    celula.fill = PatternFill(start_color="4F81BD", end_color="4F81BD", fill_type="solid")

# Salvando o arquivo
wb.save("chamados_suporte.xlsx")
print("Planilha chamados_suporte.xlsx gerada!")
```

Leitura de dados

```python
from openpyxl import load_workbook

# Carregando a planilha existente
wb = load_workbook("chamados_suporte.xlsx")
ws = wb["Relatorio_TI"]

print("--- Chamados com Prioridade Crítica ou Alta ---")
# iter_rows a partir da linha 2 para pular o cabeçalho
for linha in ws.iter_rows(min_row=2, values_only=True):
    id_chamado, sistema, prioridade, tempo = linha
    if prioridade in ["Alta", "Crítica"]:
        print(f"Alerta: Chamado {id_chamado} ({sistema}) requer atenção imediata!")
```

# XML

Exemplo

```python
import xml.etree.ElementTree as ET

# 1. Carrega e analisa o arquivo XML
arvore = ET.parse("exemplo_nfe.xml")
raiz = arvore.getroot()

# 2. Navegação direta ou busca por tags com .find() e .findall()
# Acessando dados do emitente e valor
numero_nfe = raiz.find(".//nNF").text
emitente = raiz.find(".//emit/xNome").text
valor_total = raiz.find(".//vNF").text

print(f"NF: {numero_nfe} | Emitente: {emitente} | Total: R$ {valor_total}")
```

Gerador de XML

```python
# Script para criar 3 notas fiscais simuladas no Colab:
import os

os.makedirs("notas_xml", exist_ok=True)

notas_amostra = [
    {"num": "1001", "data": "2026-10-10", "fornecedor": "Tech Solutions LTDA", "cnpj": "12.345.678/0001-90", "total": "4500.00"},
    {"num": "1002", "data": "2026-10-12", "fornecedor": "Cloud Cloud Servicos", "cnpj": "98.765.432/0001-11", "total": "1280.50"},
    {"num": "1003", "data": "2026-10-15", "fornecedor": "Hardware & Redes SA", "cnpj": "45.123.890/0001-55", "total": "8900.00"}
]

for n in notas_amostra:
    conteudo = f"""<?xml version="1.0" encoding="UTF-8"?>
<nfeProc xmlns="http://www.portalfiscal.inf.br/nfe">
    <NFe>
        <infNFe>
            <ide>
                <nNF>{n['num']}</nNF>
                <dhEmi>{n['data']}T14:30:00-03:00</dhEmi>
            </ide>
            <emit>
                <CNPJ>{n['cnpj']}</CNPJ>
                <xNome>{n['fornecedor']}</xNome>
            </emit>
            <total>
                <ICMSTot>
                    <vNF>{n['total']}</vNF>
                </ICMSTot>
            </total>
        </infNFe>
    </NFe>
</nfeProc>"""
    with open(f"notas_xml/nfe_{n['num']}.xml", "w", encoding="utf-8") as f:
        f.write(conteudo)

print("Amostras criadas na pasta 'notas_xml'!")
```
